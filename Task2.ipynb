```python
#2 ans
num=int(input("Enter a number:"))
original_num=num
num_digits=len(str(num))
digit_sum=0
while num>0:
  digit=num%10
  digit_sum+=digit**num_digits
  num//=10
if digit_sum==original_num:
  print(original_num,"is an armstrong number")
else:
  print(original_num, "is not an armstrong number")
```

```python
# function
#1 ans
def unique_element(first_list):
    unique_list=[]
    for element in first_list:
        if element not in unique_list:
           unique_list.append(element)
    return unique_list
sample_list=[1,2,3,3,3,4,4,5]
print("Unique List:",unique_element(sample_list))
```

```python
#2 ans
def count_vowels(string):
  vowels="aeiouAEIOU"
  count=0
  for char in string:
      if char in vowels:
         count+=1
  return count
print("Vowels Count:",count_vowels("photography"))
```

```python
#3 ans
def prime(n):
  if n<=1:
    return False
  for i in range(2,int(n**0.5)+1):
    if n%i==0:
      return False
  return True
print("Is 13 prime?",prime(13))
print("is 6 prime?",prime(6))
```

```python
#4 ans
def cal():
  print("simple calculator")
  print("operation:+,-,*,/")
  num1=int(input("Enter first number:"))
  operation=input("Enter operations(+,-,*,/):")
  num2=int(input("Enter second number:"))
  if operation=='+':
      result=num1+num2
  elif operation=='-':
      result=num1-num2
  elif operation=='*':
      result=num1*num2
  elif operation=='/':
    if num2!=0:
      result=num1/num2
    else:
      result="Not Division by zero "
  else:
    result="invalid operation selected"
  print("Result:",result)
cal()
```
