```python
#list program
obj=["chair","table","fan","table"]
obj[3]="remote"
print(obj)
```

```python
#append program
veg=["carrot","beans","tomato"]
veg.append("onion")
print(veg)
```

```python
# extend program
veg=["carrot","beans"]
veg.extend(["tomato","potato","onion"])
print(veg)
```

```python
# remove program
num=["1","2","3","3","4","5"]
num.remove("3")
print(num)
```

```python
# pop program
car=["bmw","toyato","audi"]
car.pop(1)
print(car)
```

```python
#clear program
fruit=["apple","banana","orange"]
fruit.clear()
print(fruit)
```

```python
#sort program ass order
num=["43","32","35","29","59"]
num.sort()
print(num)
```

```python
#reverse program
num=["43","32","35","29","59"]
num.reverse()
print(num)
```

```python
# sort reverse dec order
num=["43","32","35","29","59"]
num.sort(reverse=True)
print(num)
```

```python
#slicing program
List=["89","30","67","53","10","71","2"]
print("1",List[:3])
print("2",List[0:4])
print("3",List[-4:-1])
print("4",List[::2])
print("5",List[-2])
print("6",List[2:7])
print("7",List[::3])
print("8",List[3:7])
print("9",List[::1])
print("10",List[2:5:2])
```

```python
#concatenate program(add)
a=(1,2,3)
b=(4,5,6)
Concatenate=a+b
print(Concatenate)

week1="sun","mon","tue"
week2="wed","thurs","fri","sat"
weeks=week1+week2
print(weeks)
```

```python
#count program
fruits=("apple","banana","grapes","orange","kiwi","apple","dragon fruit")
print(fruits.count("apple"))
```

```python
#index program
fruits=("apple","banana","grapes","orange","kiwi","apple","dragon fruit")
print(fruits.index("apple"))
```

```python
#set update
a={"mon","tue","wed","thurs","fri"}
a.update(["sat"])
print(a)
```

```python
#set remove,discard,pop & clear
a={"Mon","Tue","Wed","Thurs","Fri"}
a.update(["Sat"])
print(a)
a.remove("Sat")
print(a)
a.discard("Thurs")
print(a)
a.pop()
print(a)
a.clear()
print(a)
```

```python
# dictiory
clg={
    "event":"farewall",
    "name":"sri",
    "dresscode":"formal",
    "branch":"cs",
    }
#name check
print(clg["name"])

#insert/create new key&value
clg["place"]="Chennai"
print(clg)

#update
clg["branch"]="DS"
print(clg)

#pop
clg.pop("place")
print(clg)

#delete
del clg["branch"]
print(clg)

#clear
clg.clear()
print(clg)
```
