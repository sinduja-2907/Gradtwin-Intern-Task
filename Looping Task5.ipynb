```python
#20 ans
numbers=[10,20,30,0,-80,-77,-45,0]
positive=0
negative=0
zero=0
for i in numbers:
  if i>0:
    positive=positive+1
  elif i<0:
    negative=negative+1
  else:
    zero=zero+1
print("Positive:",positive)
print("Negative:",negative)
print("Zero:",zero)
```

```python
#27 ans
num=int(input("Enter a number:"))
reverse=0
while num>0:
  digit=num%10
  reverse=reverse*10+digit
  num=num//10
print("Reverse:",reverse)
```

```python
#29 ans
num=int(input("Enter a number:"))
count=0
while num>0:
  num=num//10
  count=count+1
print("number of digit:",count)
```

```python
#30 ans
num=int(input("Enter a number:"))
temp=num
reverse=0
while temp>0:
  digit=temp%10
  reverse=reverse*10+digit
  temp//=10
if num==reverse:
  print("Palindrome")
else:
  print("Not palindrome")
```
