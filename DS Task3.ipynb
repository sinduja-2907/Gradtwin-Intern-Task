```python
#38 ans
students={
    "abi":85,
    "ravi":33,
    "sri":31,
    "diya":40
}
pass_students={}
fail_students={}
for name in students:
  if students[name]>=35:
    pass_students[name]=students[name]
  else:
    fail_students[name]=students[name]
print("Pass:",pass_students)
print("Fail:",fail_students)
```

```python
#39 ans
phonebook = {}
while True:
    print("\n1. Add")
    print("2. Search")
    print("3. Update")
    print("4. Delete")
    print("5. Exit")
    choice = int(input("Enter choice: "))
    if choice == 1:
        name = input("Enter name: ")
        phone = input("Enter phone: ")
        phonebook[name] = phone
        print("Contact added")
    elif choice == 2:
        name = input("Enter name: ")
        if name in phonebook:
            print("Phone:", phonebook[name])
        else:
            print("Contact not found")
    elif choice == 3:
        name = input("Enter name: ")
        if name in phonebook:
            phone = input("Enter new phone: ")
            phonebook[name] = phone
            print("Contact updated")
        else:
            print("Contact not found")
    elif choice == 4:
        name = input("Enter name: ")
        if name in phonebook:
            del phonebook[name]
            print("Contact deleted")
        else:
            print("Contact not found")
    elif choice == 5:
        break
    else:
        print("Invalid choice")
```
